```python
# Create a function that tells elegibility of marriage for male and female according to thei
r age limit like 21 for male and 18 for female
```

```python
def elegibilityMarriage():
    gender = input("Your Gender:")
    marriageAge = int(input("Your Age:"))
    
    if gender =="Male" and marriageAge >=21 :
       print ("ELIGIBLE")
    if gender =="Female" and marriageAge >=18 :
       print ("ELIGIBLE")
    else:
       print ("Not ELIGIBLE")  
```

```python
elegibilityMarriage()
```
