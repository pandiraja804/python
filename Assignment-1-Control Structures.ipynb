```python
bmiInputValue = float(input("Enter the BMI Index"))
```

```python
if(bmiInputValue <18.5):
    print("Underweight")
elif(bmiInputValue <24.9):
    print("Healthy Weight")
elif(bmiInputValue<29.9):
    print("Overweight")
elif(bmiInputValue<34.9):
    print("Obese Class 1 or Very Overweight")
elif(bmiInputValue<39.9):
    print("Obese Class 2")    
else:
    print("Obese Class 3")
    
```
